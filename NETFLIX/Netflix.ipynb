```python
# ==========================================
# CELL 1 - IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import os
import glob
import warnings

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# ==========================================
# CELL 2 - CHECK JUPYTER FOLDER
# ==========================================

print("Current working directory:")
print(os.getcwd())

print("\nFiles available in this folder:")

for file in os.listdir():
    print(file)
```

```python
# ==========================================
# CELL 3 - FIND CSV FILE
# ==========================================

csv_files = glob.glob("*.csv")

print("CSV files found:")

for file in csv_files:
    print(file)

if len(csv_files) == 0:
    print("\n❌ No CSV file found!")
else:
    print("\n✅ CSV file found!")
```

```python
# ==========================================
# CELL 4 - LOAD DATASET
# ==========================================

if len(csv_files) == 0:

    print("❌ No CSV file found.")

elif len(csv_files) == 1:

    file_path = csv_files[0]

    df = pd.read_csv(file_path)

    print("✅ Dataset loaded successfully!")
    print("File:", file_path)

else:

    print("Multiple CSV files found:")

    for i, file in enumerate(csv_files):
        print(i, "-", file)

    # Select first CSV
    file_path = csv_files[0]

    df = pd.read_csv(file_path)

    print("\nUsing:", file_path)
```

```python
# ==========================================
# CELL 5 - BASIC DATASET INFORMATION
# ==========================================

print("Dataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# ==========================================
# CELL 6 - FIRST 10 RECORDS
# ==========================================

display(df.head(10))
# ==========================================
# CELL 7 - DATA TYPES
# ==========================================

df.info()
# ==========================================
# CELL 8 - STATISTICAL SUMMARY
# ==========================================

display(df.describe(include="all"))
```

```python
# ==========================================
# CELL 9 - MISSING VALUES
# ==========================================

missing = df.isnull().sum()

print("Missing values:")
print(missing)

print("\nTotal missing values:")
print(missing.sum())
# ==========================================
# CELL 10 - MISSING VALUE PERCENTAGE
# ==========================================

missing_percentage = (
    df.isnull().sum()
    / len(df)
    * 100
)

missing_table = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": missing_percentage.round(2)
})

display(
    missing_table.sort_values(
        "Missing_Percentage",
        ascending=False
    )
)
# ==========================================
# CELL 11 - DUPLICATES
# ==========================================

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
# ==========================================
# REMOVE DUPLICATES
# ==========================================

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
```

```python
# ==========================================
# CELL 9 - MISSING VALUES
# ==========================================

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:")
print(missing_values.sum())
# ==========================================
# CELL 10 - DUPLICATE ROWS
# ==========================================

duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)# ==========================================
# REMOVE DUPLICATES
# ==========================================

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
```

```python
# Count Movies and TV Shows

content_type = df["type"].value_counts()

print(content_type)
plt.figure(figsize=(8, 5))

sns.barplot(
    x=content_type.index,
    y=content_type.values
)

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
percentage = (
    df["type"]
    .value_counts(normalize=True)
    * 100
)

print("Content Type Percentage:")
print(percentage.round(2))
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    content_type.values,
    labels=content_type.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Netflix Content Distribution")

plt.show()
```

```python
countries = (
    df["country"]
    .dropna()
    .str.split(", ")
    .explode()
)

country_count = countries.value_counts()

print("Top 20 Countries:")
display(country_count.head(20))
plt.figure(figsize=(12, 8))

sns.barplot(
    x=country_count.head(15).values,
    y=country_count.head(15).index
)

plt.title("Top 15 Countries by Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

```python
genres = (
    df["listed_in"]
    .dropna()
    .str.split(", ")
    .explode()
)

genre_count = genres.value_counts()

print("Top 20 Genres:")
display(genre_count.head(20))
plt.figure(figsize=(12, 8))

sns.barplot(
    x=genre_count.head(15).values,
    y=genre_count.head(15).index
)

plt.title("Top 15 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

```python

ratings = df["rating"].value_counts()

print("Netflix Ratings:")
display(ratings)
plt.figure(figsize=(12, 6))

sns.barplot(
    x=ratings.head(10).values,
    y=ratings.head(10).index
)

plt.title("Top Netflix Content Ratings")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.show()
```

```python
release_year = (
    df["release_year"]
    .value_counts()
    .sort_index()
)

display(release_year)
plt.figure(figsize=(14, 6))

sns.lineplot(
    x=release_year.index,
    y=release_year.values
)

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
movies = df[df["type"] == "Movie"]

movie_year = (
    movies["release_year"]
    .value_counts()
    .sort_index()
)

display(movie_year)
plt.figure(figsize=(14, 6))

sns.lineplot(
    x=movie_year.index,
    y=movie_year.values
)

plt.title("Movies by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")

plt.show()
```

```python
tv_shows = df[df["type"] == "TV Show"]

tv_year = (
    tv_shows["release_year"]
    .value_counts()
    .sort_index()
)

display(tv_year)
plt.figure(figsize=(14, 6))

sns.lineplot(
    x=tv_year.index,
    y=tv_year.values
)

plt.title("TV Shows by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of TV Shows")

plt.show()
```

```python
directors = (
    df["director"]
    .dropna()
    .value_counts()
)

print("Top 20 Directors:")
display(directors.head(20))
plt.figure(figsize=(12, 8))

sns.barplot(
    x=directors.head(15).values,
    y=directors.head(15).index
)

plt.title("Top 15 Directors by Number of Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Director")

plt.show()
```

```python
movie_duration = (
    movies["duration"]
    .dropna()
    .str.extract(r"(\d+)")
)

movie_duration = pd.to_numeric(
    movie_duration[0],
    errors="coerce"
)

print("Movie Duration Statistics:")

print(
    movie_duration.describe()
)
plt.figure(figsize=(10, 6))

sns.histplot(
    movie_duration,
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

```python
tv_seasons = (
    tv_shows["duration"]
    .dropna()
    .str.extract(r"(\d+)")
)

tv_seasons = pd.to_numeric(
    tv_seasons[0],
    errors="coerce"
)

print("TV Show Season Statistics:")

print(
    tv_seasons.describe()
)
plt.figure(figsize=(10, 6))

sns.histplot(
    tv_seasons,
    bins=15,
    discrete=True
)

plt.title("Distribution of TV Show Seasons")
plt.xlabel("Number of Seasons")
plt.ylabel("Number of TV Shows")

plt.show()
```

```python
date_added = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

added_year = (
    date_added
    .dt.year
    .value_counts()
    .sort_index()
)

display(added_year)
plt.figure(figsize=(14, 6))

sns.lineplot(
    x=added_year.index,
    y=added_year.values,
    marker="o"
)

plt.title("Netflix Content Added Over the Years")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")

plt.show()
```

```python
added_month = (
    date_added
    .dt.month_name()
    .value_counts()
)

display(added_month)
month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

added_month = (
    date_added
    .dt.month_name()
    .value_counts()
    .reindex(month_order)
)

display(added_month)
plt.figure(figsize=(14, 6))

sns.barplot(
    x=added_month.index,
    y=added_month.values
)

plt.title("Netflix Content Added by Month")
plt.xlabel("Month")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
country_type = (
    df[
        ["country", "type"]
    ]
    .dropna()
    .assign(
        country=lambda x: x["country"].str.split(", ")
    )
    .explode("country")
)

country_type_count = (
    country_type
    .groupby(
        ["country", "type"]
    )
    .size()
    .reset_index(name="count")
)

top_countries = (
    country_type["country"]
    .value_counts()
    .head(10)
    .index
)

country_type_top = (
    country_type_count[
        country_type_count["country"].isin(top_countries)
    ]
)

display(country_type_top)
plt.figure(figsize=(14, 7))

sns.barplot(
    data=country_type_top,
    x="count",
    y="country",
    hue="type"
)

plt.title("Movies vs TV Shows in Top Countries")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

```python
genre_type = (
    df[
        ["listed_in", "type"]
    ]
    .dropna()
    .assign(
        genre=lambda x: x["listed_in"].str.split(", ")
    )
    .explode("genre")
)

genre_type_count = (
    genre_type
    .groupby(
        ["genre", "type"]
    )
    .size()
    .reset_index(name="count")
)

top_genres = (
    genre_type["genre"]
    .value_counts()
    .head(10)
    .index
)

genre_type_top = (
    genre_type_count[
        genre_type_count["genre"].isin(top_genres)
    ]
)

display(genre_type_top)
plt.figure(figsize=(14, 7))

sns.barplot(
    data=genre_type_top,
    x="count",
    y="genre",
    hue="type"
)

plt.title("Top Genres: Movies vs TV Shows")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

```python
# ==========================================
# KEY INSIGHTS
# ==========================================

total_titles = len(df)

total_movies = (
    df["type"] == "Movie"
).sum()

total_tv = (
    df["type"] == "TV Show"
).sum()

most_common_type = (
    df["type"]
    .value_counts()
    .idxmax()
)

top_country = country_count.idxmax()

top_genre = genre_count.idxmax()

top_rating = ratings.idxmax()

top_director = directors.idxmax()

average_movie_duration = movie_duration.mean()

average_tv_seasons = tv_seasons.mean()

print("============================================")
print("           NETFLIX KEY INSIGHTS")
print("============================================")

print(f"Total Netflix Titles       : {total_titles:,}")

print(f"Total Movies               : {total_movies:,}")

print(f"Total TV Shows             : {total_tv:,}")

print(f"Most Common Content Type   : {most_common_type}")

print(f"Top Content Country        : {top_country}")

print(f"Most Popular Genre         : {top_genre}")

print(f"Most Common Rating         : {top_rating}")

print(f"Top Director               : {top_director}")

print(
    f"Average Movie Duration     : "
    f"{average_movie_duration:.2f} minutes"
)

print(
    f"Average TV Show Seasons    : "
    f"{average_tv_seasons:.2f}"
)

print("============================================")
```
